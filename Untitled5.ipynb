```python

from collections import deque

# 예시 그래프: 친구 관계
graph = {
    'Alice': ['Bob', 'Charlie'],
    'Bob': ['David', 'Eve'],
    'Charlie': ['Frank'],
    'David': [],
    'Eve': ['Frank'],
    'Frank': []
}


# 1. DFS 구현
def dfs(graph, v, visited):
    visited.add(v)
    print(v, end=' ')

    for neighbor in graph[v]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)


print("DFS 탐색 순서:")
dfs(graph, 'Alice', set())


# 2. BFS 구현
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    while queue:
        v = queue.popleft()
        print(v, end=' ')

        for neighbor in graph[v]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)


print("\nBFS 탐색 순서:")
bfs(graph, 'Alice')


# 3. 친구 추천 시스템 (BFS 응용)
def recommend_friend(graph, start):
    visited = set()
    queue = deque([(start, 0)])

    print(f"\n{start}의 친구 추천 (2단계):")

    while queue:
        current, depth = queue.popleft()

        if current not in visited:
            visited.add(current)

            if depth == 2:
                print("-", current)

            if depth < 2:
                for neighbor in graph[current]:
                    queue.append((neighbor, depth + 1))


recommend_friend(graph, 'Alice')
```
